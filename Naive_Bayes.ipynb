```python
#Import Library to read Data
!pip install tweet-preprocessor
import pandas as pd
import preprocessor as p
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
#Read Extraction data

testdata=pd.read_csv('look.csv')

def preprocess_tweet(text):
    text = p.clean(text)
    return text
testdata['Tweet']=testdata['Tweet'].apply(preprocess_tweet)
testdata[['Tweet']].head()
```

```python

tokenized_tweet2=testdata['Tweet'].apply(lambda x: x.split())
tokenized_tweet2.head()
tokenized_tweet2=tokenized_tweet2.apply(lambda x:[lemmatizer.lemmatize(i) for i in x])
tokenized_tweet2.head()
stopped=['Follow','vaccine','covid','amp','amp','amp','Vaccine','people','People']
tokenized_tweet2=tokenized_tweet2.apply(lambda x: [item for item in x if item not in stopped])
tokenized_tweet2.head()
for i in range(len(tokenized_tweet2)):
    tokenized_tweet2[i]=' '.join(tokenized_tweet2[i])
testdata['Tweet']=tokenized_tweet2
testdata[['Tweet']].head()
```

# **CLEANING TEST DATA**

```python
import sklearn
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer(max_features=5000)
testvector=vectorizer.fit_transform(testdata['Tweet'])
```

```python
testvector.shape
```

```python
X_features = testvector
y=testdata['Analysis']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_features, y, test_size = 0.2, random_state = 1)
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
NaiveBclassifier = MultinomialNB()
NaiveBclassifier.fit(X_train,y_train)
y_pred=NaiveBclassifier.predict(X_test)
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(y_test,y_pred)*100
print(accuracy)
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print('Confusion matrix\n\n', cm)

print('\nTrue Positives(TP) = ', cm[0,0])

print('\nTrue Negatives(TN) = ', cm[1,1])

print('\nFalse Positives(FP) = ', cm[0,1])

print('\nFalse Negatives(FN) = ', cm[1,0])
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
